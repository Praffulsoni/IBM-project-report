# SpaceX API Data Collection
IBM Applied Data Science Capstone

REST workflow: GET request → JSON response → json_normalize() → feature selection → analysis DataFrame.

```python
import requests
import pandas as pd

url = 'https://api.spacexdata.com/v4/launches'
response = requests.get(url, timeout=30)
response.raise_for_status()
launches = response.json()
df = pd.json_normalize(launches)
df.shape
```

```python
cols = [c for c in ['flight_number','name','date_utc','success','rocket','launchpad','payloads','cores','latitude','longitude'] if c in df.columns]
api_df = df[cols].copy()
api_df.head()
```

## Result
Launch records are normalized into a tabular DataFrame for downstream wrangling and analysis.
